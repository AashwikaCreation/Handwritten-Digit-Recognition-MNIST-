```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
```

```python
# MNIST dataset loading (Train & Test splits)
(X_train, y_train), (X_test, y_test) = datasets.mnist.load_data()

# Data normalization (sclaing pixels 0 to 1 )
X_train = X_train / 255.0
X_test = X_test / 255.0

# for CNN reshaping image dimensions (28x28 to 28x28x1)
X_train = X_train.reshape((-1, 28, 28, 1))
X_test = X_test.reshape((-1, 28, 28, 1))
```

```python
# Convolutional Neural Network (CNN) model banana
model = models.Sequential([
    # Convolutional Layer
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    # Doosri Conv Layer
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Dense Layers (Classification ke liye)
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax') # 10 digits (0 to 9) ke liye output
])

model.summary()
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Model train karna (5 epochs me hi acchi accuracy aa jati hai)
history = model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Model training (get accuracy generally in 5 epochs )
history = model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"\nTest Accuracy: {test_acc * 100:.2f}%")
```
